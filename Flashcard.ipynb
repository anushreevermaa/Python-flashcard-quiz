```python
from IPython.display import clear_output
import random
import time

QUESTION_BANK = {
    "What keyword defines a function in Python?": "def",
    "What data structure uses square brackets []?": "list",
    "What function gets user input in Python 3?": "input",
    "What data type returns True or False?": "boolean",
    "What keyword stops a loop prematurely?": "break",
    "What keyword skips the rest of the current loop iteration?": "continue",
    "Which library adds random selection functionality?": "random",
    "What symbol starts a single-line comment in Python?": "#",
    "What function returns the number of items in a list?": "len",
    "What immutable sequence data type uses parentheses ()?": "tuple",
    "What method converts all text in a string to uppercase?": "upper",
    "What method converts all text in a string to lowercase?": "lower",
    "Which operator is used for exponentiation (power)?": "**",
    "What keyword is used for conditional branching?": "if",
    "What method removes leading and trailing spaces from text?": "strip",
    "What function converts a value into an integer?": "int",
    "What function converts a value into a floating-point number?": "float",
    "What built-in function displays text to the console?": "print",
    "What data type stores Key-Value pairs using {}?": "dictionary",
    "Which module handles time delays like sleep()?": "time",
    "What keyword is used to import a library?": "import",
}


def select_random_questions(question_bank, num_questions=5):
    """Randomly selects a specified number of unique questions from the bank."""
    # Grab random keys from the dictionary without duplicates
    selected_keys = random.sample(list(question_bank.keys()), num_questions)
    # Rebuild a smaller dictionary with only those 5 chosen questions
    return {key: question_bank[key] for key in selected_keys}


def run_quiz(question_bank):
    """Runs a 5-question quiz round using a randomly sampled subset."""
    # Fetch 5 random questions for this session
    quiz_cards = select_random_questions(question_bank, num_questions=5)

    score = 0
    total = len(quiz_cards)

    for question, correct_answer in quiz_cards.items():
        print("=== PYTHON FLASHCARD QUIZ ===")
        print(f"Current Score: {score}/{total}\n")
        print(f"Q: {question}")

        # Take input FIRST before clearing to prevent Jupyter kernel deadlock
        user_answer = input("Your Answer: ").strip().lower()
        clear_output()

        if user_answer == correct_answer.lower():
            print("✅ Correct!\n")
            score += 1
        else:
            print(f"❌ Incorrect. The correct answer was: '{correct_answer}'\n")

        time.sleep(1.5)
        clear_output()

    print("=== QUIZ COMPLETE ===")
    print(f"Final Score: {score}/{total} ({(score/total)*100:.0f}%)\n")


def replay():
    """Asks if the user wants to take another random round."""
    return (
        input("Would you like to try another 5 questions? (Yes/No): ")
        .lower()
        .startswith("y")
    )


# --- Main Application Loop ---
while True:
    run_quiz(QUESTION_BANK)
    if not replay():
        clear_output()
        print("Great job studying! Keep it up!")
        break
```
